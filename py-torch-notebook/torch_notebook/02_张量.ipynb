```python
import torch
```

```python
x = torch.rand(4, 3) # 随机初始化一个矩阵
print(x)
```

```python
x = torch.zeros(4, 3, dtype=torch.long) # 构建一个全0矩阵
print(x)
```

```python
# 直接构建张量
x = torch.Tensor([5.5, 3])
print(x)
```

```python
x = x.new_ones(4, 3, dtype=torch.double) 
# 创建一个新的全1矩阵tensor，返回的tensor默认具有相同的torch.dtype和torch.device
# 也可以像之前的写法 x = torch.ones(4, 3, dtype=torch.double)
print(x)
x = torch.randn_like(x, dtype=torch.float)
# 重置数据类型
print(x)
# 结果会有一样的size
# 获取它的维度信息
print(x.size())
print(x.shape)
```

## 张量操作

```python
# 加法操作
y = torch.rand(4, 3)
print(y.size())
print(y)
print(50 * "*")
print(x)
print(50 * "*")
print(x + y)
print(50 * "*")
print(y.add_(x))
```

```python
# 索引操作
x = torch.rand(4, 3)
# 取第2列
print(x)
print(50 * "-")
print(x[:, 1])
print(50 * "-")
# 需要注意的是：索引出来的结果与原数据共享内存，修改一个，另一个会跟着修改。如果不想修改，可以考虑使用copy()等方法
y = x[0, :]
y += 1
print(y)
print(50 * "-")
print(x[0, :])
```

```python
# 维度变换
x = torch.randn(4, 4)
y = x.view(16)
z = x.view(-1, 8) # -1指这一维度的维数又其他维度决定
print(x.size(), y.size(), z.size())
```
