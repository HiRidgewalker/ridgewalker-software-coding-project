## 理论知识
- 混淆矩阵  
  混淆矩阵是评估分类模型性能的重要工具，它的作用包括:  
  1. 直观展示预测结果
    - 以矩阵形式展示真实标签与预测标签的对应关系
    - 清晰显示模型在哪里做对了，哪里做错了
  2. 计算多种评价指标
    - 准确率: (TP + TN) / 总数
    - 精确率: TP / (TP + FP) - 预测为正的样本中真正为正的比例
    - 召回率: TP / (TP + FN) - 真正为正的样本被正确预测的比例
    - F1分数: 精确率和召回率的调和平均
  3. 识别模型问题
    - 可以发现模型是否偏向某一类（类别不平衡问题）
    - 识别模型经常混淆的类别

  二分类混淆矩阵结构  
  |          | 预测为正    | 预测为负    |
  |----------|-------------|-------------|
  | 实际为正 | TP (真阳性) | FN (假阴性) |
  | 实际为负 | FP (假阳性) | TN (真阴性) |

  实际应用场景  
  - 医疗诊断: 召回率更重要（漏诊风险高）
  - 垃圾邮件分类: 精确率更重要（误判正常邮件代价高）
  - 异常检测: 关注特定类型的错误  
  混淆矩阵特别适合处理类别不平衡的数据集，单纯用准确率可能会产生误导。
