```python
import numpy as np
```

```python
print(np.__version__)
```

```python
a = np.array([1, 2, 3])
```

```python
print(a)
print(f"type: {type(a)}")
```

```python
b = np.array([[1, 2], [3, 4]])
print(b)
print(f"type: {type(b)}")
```

```python
# 指定生成数组的最小维度
c = np.array([1, 2, 3], ndmin=2)
print(c)
```

```python
# 指定类型
d = np.array([1, 2, 3], dtype=complex)
print(d)
```

```python
# ndarray的属性
print(b)
print(b.ndim)    # nda的秩
print(b.shape)  # nda的形状
print(b.dtype)  # nda的元素类型
print(b.itemsize)   # nda中每个元素的大小
```

```python
# 调整nda形状
print(b)
b1 = b.reshape(1, 4)
print(b1)
```

```python
# 创建特定类型的数组
x1 = np.zeros(5)
print(x1)
x2 = np.zeros((2, 2))
print(x2)
x3 = np.ones([2, 2])
print(x3)

x4 = np.array([[1, 2, 3], [4, 5, 6]])
print(x4)
print(np.zeros_like(x4))
print(np.ones_like(x4))
```

```python
# 将列表转换为nda
x1 = [1, 2, 3]
x2 = np.asarray(x1)
print(x2)
print(type(x2))

# 将元组转换为nda
x3 = (1, 2, 3)
x4 = np.asarray(x3)
print(x4)
print(type(x4))
```

```python
# 通过范围创建数组
# 默认: start 0; step 1
x1 = np.arange(5)
print(x1)

# x2 = np.linspace(1, 10, 10, retstep=True)
x2 = np.linspace(1, 10, 10, retstep=False)
"""
numpy.linspace(
    start, 	序列的起始值
    stop,	序列的终止值，如果endpoint为true，该值包含于数列中
    num,	要生成的等步长的样本数量，默认为50
    endpoint,	该值为 true 时，数列中包含stop值，反之不包含，默认是True。
    retstep,	如果为 True 时，生成的数组中会显示间距，反之不显示。
    dtype	ndarray 的数据类型
)
"""
print(x2)
```
